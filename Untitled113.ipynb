```python
# ==========================================
# CELL 1: Imports & Global Configuration
# ==========================================
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import time
import textwrap

from sklearn.datasets import make_moons, make_circles, make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier

# --- Research Configuration ---
SEED        = 42
N_RUNS      = 5          # จำนวนรอบสุ่มเพื่อหา Mean/Std
POP_SIZE    = 60         # ขนาดประชากร
GENERATIONS = 150        # จำนวนเจเนอเรชันสูงสุด
N_WEIGHTS   = 337        # จำนวนพารามิเตอร์ 2 -> 16 -> 16 -> 1
L_BOUNDS    = [(0,48), (48,320), (320,337)] # ขอบเขตของ Layer 1, 2, 3

np.random.seed(SEED)
print("Cell 1: Setup completed.")
```

```python
# ==========================================
# CELL 2: Datasets Definition
# ==========================================
def make_multixor(n_samples=800, grid=4, noise=0.25, random_state=42):
    """สร้างโจทย์ Multi-XOR (16-quadrant checkerboard)"""
    np.random.seed(random_state)
    X  = np.random.uniform(-2, 2, (n_samples, 2))
    X += np.random.randn(n_samples, 2) * noise
    xi = ((X[:,0]+2) / (4/grid)).astype(int).clip(0, grid-1)
    yi = ((X[:,1]+2) / (4/grid)).astype(int).clip(0, grid-1)
    return X, ((xi + yi) % 2).astype(float)

def prepare_data(X, y, test_size=0.2, seed=SEED):
    """แบ่ง Train/Test และทำ Data Standardization"""
    Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=test_size, random_state=seed)
    sc  = StandardScaler()
    Xtr = sc.fit_transform(Xtr)
    Xte = sc.transform(Xte)
    return Xtr, Xte, ytr, yte

def build_datasets():
    datasets = {}

    # 1. Moon
    Xm, ym = make_moons(n_samples=500, noise=0.2, random_state=SEED)
    datasets["Moon"] = prepare_data(Xm, ym)

    # 2. Multi-XOR
    Xx, yx = make_multixor()
    datasets["Multi-XOR"] = prepare_data(Xx, yx)

    # 3. Circles
    Xc, yc = make_circles(n_samples=500, noise=0.1, factor=0.4, random_state=SEED)
    datasets["Circles"] = prepare_data(Xc, yc)

    # 4. Blobs
    Xb, yb = make_classification(n_samples=500, n_features=2, n_informative=2,
                                 n_redundant=0, n_clusters_per_class=2, random_state=SEED)
    datasets["Blobs"] = prepare_data(Xb, yb.astype(float))

    print(f"Cell 2: Datasets built. Total: {len(datasets)} datasets.")
    return datasets

# รันเก็บค่าไว้ใน memory เลย
DATASETS = build_datasets()
```

```python
# ==========================================
# CELL 3: Neural Network Core & Utilities
# ==========================================
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-np.clip(x, -500, 500)))

def nn_forward(X, w):
    """Forward pass 2 -> 16 -> 16 -> 1 (337 weights)"""
    W1 = w[0:32].reshape(2,16);    b1 = w[32:48]
    W2 = w[48:304].reshape(16,16); b2 = w[304:320]
    W3 = w[320:336].reshape(16,1); b3 = w[336:337]

    h1 = relu(X @ W1 + b1)
    h2 = relu(h1 @ W2 + b2)
    return sigmoid(h2 @ W3 + b3).flatten()

def compute_loss(w, X, y):
    """Binary Cross-Entropy Loss"""
    p = np.clip(nn_forward(X, w), 1e-9, 1-1e-9)
    return -np.mean(y * np.log(p) + (1 - y) * np.log(1 - p))

def compute_acc(w, X, y):
    """Classification Accuracy"""
    return np.mean((nn_forward(X, w) >= 0.5) == y)

def tournament_select(pop, fitness, k=5):
    """สุ่มคัดเลือกพ่อแม่พันธุ์ (Tournament Selection)"""
    idx = np.random.choice(len(pop), k, replace=False)
    return pop[idx[np.argmax(fitness[idx])]]

def alpha_schedule(gen, gens, start=0.5, end=0.01):
    """ลดขนาด Mutation Rate แบบ Exponential"""
    return start * (end/start) ** (gen/max(gens-1, 1))

# --- เครื่องมือสำหรับ DELS และ DELCMASE ---
GR = (np.sqrt(5)-1)/2  # อัตราส่วนทองคำ (Golden Ratio)

def golden_section_search(loss_fn, a=0.0, b=1.0, tol=1e-3, max_iter=10):
    """หาจุดต่ำสุด (Local Minimum) บนเส้นตรง 1 มิติ (Line Search)"""
    c  = b - GR*(b-a)
    d  = a + GR*(b-a)
    fc = loss_fn(c)
    fd = loss_fn(d)

    for _ in range(max_iter):
        if abs(b-a) < tol: break
        if fc < fd:
            b=d; d=c; fd=fc; c=b-GR*(b-a); fc=loss_fn(c)
        else:
            a=c; c=d; fc=fd; d=a+GR*(b-a); fd=loss_fn(d)

    return (a+b)/2, min(fc, fd)

print("Cell 3: Neural Net and Utilities loaded.")
```

```python
# ==========================================
# CELL 4: Baselines (Backprop, Standard GA, Standard CMA-ES)
# ==========================================

# ── 1. Backpropagation (Adam) ───────────────────────────────
def run_backprop(Xtr, ytr, Xte, yte, seed=SEED):
    t0  = time.time()
    mlp = MLPClassifier(hidden_layer_sizes=(16,16), activation='relu',
                        solver='adam', max_iter=200,
                        learning_rate_init=0.01, random_state=seed)
    mlp.fit(Xtr, ytr)
    return mlp.score(Xte, yte), time.time()-t0, getattr(mlp, 'loss_curve_', [])

# ── 2. Standard GA ──────────────────────────────────────────
def standard_ga(Xtr, ytr, Xte, yte, seed=SEED):
    np.random.seed(seed)
    pop = np.random.randn(POP_SIZE, N_WEIGHTS) * 0.1
    best_w, best_loss = None, np.inf
    hist_te, hist_loss = [], []
    t0 = time.time()

    for gen in range(GENERATIONS):
        losses  = np.array([compute_loss(ind, Xtr, ytr) for ind in pop])
        fitness = -losses
        idx_b   = np.argmin(losses)

        if losses[idx_b] < best_loss:
            best_loss = losses[idx_b]
            best_w = pop[idx_b].copy()

        hist_te.append(compute_acc(best_w, Xte, yte))
        hist_loss.append(best_loss)

        alpha_t = alpha_schedule(gen, GENERATIONS)
        new_pop = [best_w.copy()]  # Elitism

        while len(new_pop) < POP_SIZE:
            p1 = tournament_select(pop, fitness)
            p2 = tournament_select(pop, fitness)
            pt = np.random.randint(1, N_WEIGHTS)
            child = np.concatenate([p1[:pt], p2[pt:]])

            mask = np.random.rand(N_WEIGHTS) < 0.1
            child[mask] += np.random.randn(mask.sum()) * alpha_t
            new_pop.append(child)

        pop = np.array(new_pop)

    return best_w, hist_te, hist_loss, time.time()-t0

import cma

# ── Standard CMA-ES (Official Implementation) ──────────────
def official_cma_es(Xtr, ytr, Xte, yte, seed=SEED):
    np.random.seed(seed)
    # เริ่มต้นค้นหาด้วย mean 0 และ global step-size (sigma) ที่ 0.1
    es = cma.CMAEvolutionStrategy(N_WEIGHTS * [0.0], 0.1,
                                  {'popsize': POP_SIZE, 'seed': seed, 'verbose': -9})

    best_w, best_loss = None, np.inf
    hist_te, hist_loss = [], []
    t0 = time.time()

    for gen in range(GENERATIONS):
        # 1. Ask: สร้างประชากรจาก Multivariate Normal ของจริง
        solutions = es.ask()

        # 2. Evaluate: คำนวณ Loss
        losses = [compute_loss(w, Xtr, ytr) for w in solutions]

        # 3. Tell: อัปเดต Evolution paths, Step-size และ Covariance Matrix
        es.tell(solutions, losses)

        idx_best = np.argmin(losses)
        if losses[idx_best] < best_loss:
            best_loss = losses[idx_best]
            best_w    = solutions[idx_best].copy()

        hist_te.append(compute_acc(best_w, Xte, yte))
        hist_loss.append(best_loss)

    return best_w, hist_te, hist_loss, time.time()-t0
```

```python
# ==========================================
# CELL 5: Existing Hybrid Approaches (GI-GA, LAME v1)
# ==========================================

# ── 1. Gradient-Informed GA (GI-GA) ─────────────────────────
def gradient_informed_ga(Xtr, ytr, Xte, yte, seed=SEED):
    np.random.seed(seed)
    pop = np.random.randn(POP_SIZE, N_WEIGHTS) * 0.1
    best_w, best_loss = None, np.inf
    hist_te, hist_loss = [], []
    current_grad = None
    t0 = time.time()

    for gen in range(GENERATIONS):
        alpha_t = alpha_schedule(gen, GENERATIONS, 0.3, 0.01)
        beta_t  = alpha_schedule(gen, GENERATIONS, 0.5, 0.1)
        losses  = np.array([compute_loss(ind, Xtr, ytr) for ind in pop])
        fitness = -losses
        idx_b   = np.argmin(losses)

        if losses[idx_b] < best_loss:
            best_loss=losses[idx_b]; best_w=pop[idx_b].copy()

        hist_te.append(compute_acc(best_w, Xte, yte))
        hist_loss.append(best_loss)

        # Finite difference gradient ทุกๆ 10 gen บนข้อมูลสุ่ม
        if gen % 10 == 0:
            eps  = 1e-4
            sub  = np.random.choice(len(Xtr), min(40, len(Xtr)), replace=False)
            grad = np.zeros(N_WEIGHTS)
            for i in range(N_WEIGHTS):
                wp = best_w.copy(); wp[i] += eps
                wm = best_w.copy(); wm[i] -= eps
                grad[i] = (compute_loss(wp, Xtr[sub], ytr[sub]) -
                           compute_loss(wm, Xtr[sub], ytr[sub])) / (2*eps)
            current_grad = grad

        new_pop = [best_w.copy()]
        while len(new_pop) < POP_SIZE:
            p1 = tournament_select(pop, fitness)
            p2 = tournament_select(pop, fitness)
            pt = np.random.randint(1, N_WEIGHTS)
            child = np.concatenate([p1[:pt], p2[pt:]])

            # ใช้ Gradient นำทางให้กับประชากรกลุ่มแรกๆ
            if len(new_pop) < POP_SIZE//4 and current_grad is not None:
                child += -alpha_t*current_grad + beta_t*np.random.randn(N_WEIGHTS)
            else:
                mask = np.random.rand(N_WEIGHTS) < 0.1
                child[mask] += np.random.randn(mask.sum()) * beta_t
            new_pop.append(child)
        pop = np.array(new_pop)

    return best_w, hist_te, hist_loss, time.time()-t0

# ── 2. Lifespan-Aware Mutation Evolution (LAME v1) ──────────
def lame_v1(Xtr, ytr, Xte, yte, seed=SEED, age_threshold=20, alpha=0.6, beta=0.2):
    np.random.seed(seed)
    pop  = np.random.randn(POP_SIZE, N_WEIGHTS) * 0.1
    ages = np.zeros((POP_SIZE, N_WEIGHTS), dtype=int)
    prev_losses = np.full(POP_SIZE, np.inf)
    best_w, best_loss = None, np.inf
    hist_te, hist_loss = [], []
    t0 = time.time()

    for gen in range(GENERATIONS):
        losses  = np.array([compute_loss(ind, Xtr, ytr) for ind in pop])
        fitness = -losses
        idx_b   = np.argmin(losses)

        if losses[idx_b] < best_loss:
            best_loss=losses[idx_b]; best_w=pop[idx_b].copy()

        hist_te.append(compute_acc(best_w, Xte, yte))
        hist_loss.append(best_loss)

        # อัปเดตอายุ
        for i in range(POP_SIZE):
            if losses[i] < prev_losses[i]: ages[i] = np.maximum(0, ages[i]-2)
            else: ages[i] += 1
        prev_losses = losses.copy()

        top_k = max(5, POP_SIZE//4)
        top_idx = np.argsort(losses)[:top_k]
        top_mean = pop[top_idx].mean(axis=0)
        alpha_t = alpha_schedule(gen, GENERATIONS, alpha, 0.05)
        beta_t  = alpha_schedule(gen, GENERATIONS, beta,  0.5)

        # ซ่อมแซมน้ำหนักที่แก่เกินไป
        for i in range(POP_SIZE):
            old = ages[i] > age_threshold
            if old.sum() == 0: continue
            pop[i, old] += (alpha_t*(top_mean-pop[i])[old] + beta_t*np.random.randn(old.sum()))
            ages[i, old] = 0

        new_pop  = [best_w.copy()]
        new_ages = [np.zeros(N_WEIGHTS, dtype=int)]

        while len(new_pop) < POP_SIZE:
            p1i = np.random.choice(top_idx)
            p2i = np.random.choice(POP_SIZE)
            child      = pop[p1i].copy()
            child_ages = ages[p1i].copy()
            for s, e in L_BOUNDS:
                if np.random.rand() < 0.5:
                    child[s:e] = pop[p2i, s:e]
                    child_ages[s:e] = ages[p2i, s:e]
            mask = np.random.rand(N_WEIGHTS) < 0.05
            child[mask] += np.random.randn(mask.sum()) * beta_t * 0.5
            new_pop.append(child)
            new_ages.append(child_ages)

        pop = np.array(new_pop); ages = np.array(new_ages)

    return best_w, hist_te, hist_loss, time.time()-t0

print("Cell 5: Existing Hybrids loaded.")
```

```python
# ==========================================
# CELL 6: Proposed Novel Algorithms (MEGA, DELS, DELCMASE)
# ==========================================

# ── 1. MEGA (Momentum-guided Evolutionary Gradient) ──────────
def mega(Xtr, ytr, Xte, yte, seed=SEED, beta_mom=0.9, alpha_mom=0.3, gamma_lin=0.4, eps_rand=0.2):
    np.random.seed(seed)
    pop      = np.random.randn(POP_SIZE, N_WEIGHTS) * 0.1
    momentum = np.zeros(N_WEIGHTS)
    best_w, prev_best = pop[0].copy(), pop[0].copy()
    best_loss = np.inf
    hist_te, hist_loss = [], []
    t0 = time.time()

    for gen in range(GENERATIONS):
        losses  = np.array([compute_loss(ind, Xtr, ytr) for ind in pop])
        fitness = -losses
        idx_b   = np.argmin(losses)

        if losses[idx_b] < best_loss:
            best_loss=losses[idx_b]; best_w=pop[idx_b].copy()

        hist_te.append(compute_acc(best_w, Xte, yte))
        hist_loss.append(best_loss)

        delta_best = best_w - prev_best
        momentum   = beta_mom*momentum + (1-beta_mom)*delta_best
        prev_best  = best_w.copy()
        m_corr     = momentum / (1 - beta_mom**(gen+1))

        alpha_t = alpha_schedule(gen, GENERATIONS, alpha_mom, 0.05)
        gamma_t = alpha_schedule(gen, GENERATIONS, gamma_lin, 0.1)
        eps_t   = alpha_schedule(gen, GENERATIONS, eps_rand,  0.05)

        new_pop = [best_w.copy()]
        while len(new_pop) < POP_SIZE:
            p1 = tournament_select(pop, fitness)
            p2 = tournament_select(pop, fitness)
            pt = np.random.randint(1, N_WEIGHTS)
            child = np.concatenate([p1[:pt], p2[pt:]])
            child += (alpha_t*m_corr + gamma_t*(best_w - child) + eps_t*np.random.randn(N_WEIGHTS))
            new_pop.append(child)
        pop = np.array(new_pop)

    return best_w, hist_te, hist_loss, time.time()-t0

# ── 2. DELS (Direction Evolution with Line Search) ───────────
def dels(Xtr, ytr, Xte, yte, seed=SEED, pop_size=30, alpha_max=2.0, mut_scale=0.3):
    np.random.seed(seed)
    w_current  = np.random.randn(N_WEIGHTS) * 0.1
    directions = np.random.randn(pop_size, N_WEIGHTS)
    directions /= (np.linalg.norm(directions, axis=1, keepdims=True) + 1e-8)

    best_w, best_loss = w_current.copy(), compute_loss(w_current, Xtr, ytr)
    hist_te, hist_loss = [], []
    t0 = time.time()

    for gen in range(GENERATIONS):
        ls_losses = np.zeros(pop_size)
        ls_alphas = np.zeros(pop_size)

        for i, d in enumerate(directions):
            def loss_along(alpha, _d=d): return compute_loss(w_current + alpha*_d, Xtr, ytr)
            a_star, l_star = golden_section_search(loss_along, 0.0, alpha_max)
            ls_losses[i] = l_star; ls_alphas[i] = a_star

        idx_best = np.argmin(ls_losses)
        w_cand   = w_current + ls_alphas[idx_best]*directions[idx_best]

        if ls_losses[idx_best] < best_loss:
            best_loss = ls_losses[idx_best]
            w_current = w_cand.copy()
            best_w    = w_current.copy()

        hist_te.append(compute_acc(best_w, Xte, yte))
        hist_loss.append(best_loss)

        fitness = -ls_losses
        mut_t   = mut_scale * (0.05/mut_scale)**(gen/max(GENERATIONS-1,1))
        new_dirs = [directions[idx_best].copy()]
        while len(new_dirs) < pop_size:
            p1 = tournament_select(directions, fitness)
            p2 = tournament_select(directions, fitness)
            lam   = np.random.rand()
            child = lam*p1 + (1-lam)*p2
            child += np.random.randn(N_WEIGHTS) * mut_t
            child /= (np.linalg.norm(child) + 1e-8)
            new_dirs.append(child)
        directions = np.array(new_dirs)

    return best_w, hist_te, hist_loss, time.time()-t0

# ── 3. DELCMASE (The New Hybrid State-of-the-Art) 🌟 ─────────
def delcmase(Xtr, ytr, Xte, yte, seed=SEED, pop_size=30, alpha_max=2.0):
    """วิวัฒนาการทิศทางด้วย Covariance Matrix แล้ววิ่ง Line Search หาจุดสิ้นสุด"""
    np.random.seed(seed)
    w_current = np.random.randn(N_WEIGHTS) * 0.1
    best_w, best_loss = w_current.copy(), compute_loss(w_current, Xtr, ytr)

    # 🌟 สร้างจุดศูนย์กลางและความแปรปรวนสำหรับ "ทิศทาง" (Directions)
    mean_d = np.zeros(N_WEIGHTS)
    cov_d  = np.eye(N_WEIGHTS) * 0.5

    hist_te, hist_loss = [], []
    t0 = time.time()

    for gen in range(GENERATIONS):
        # 1. สุ่มทิศทางจากความรู้ Covariance ของรุ่นก่อน
        directions = np.random.multivariate_normal(mean_d, cov_d, pop_size)
        directions /= (np.linalg.norm(directions, axis=1, keepdims=True) + 1e-8) # ปรับเป็น Unit Vector

        ls_losses = np.zeros(pop_size)
        ls_alphas = np.zeros(pop_size)

        # 2. Line Search ยิงหาจุดต่ำสุดตามแนวทิศทาง
        for i, d in enumerate(directions):
            def loss_along(alpha, _d=d): return compute_loss(w_current + alpha*_d, Xtr, ytr)
            a_star, l_star = golden_section_search(loss_along, 0.0, alpha_max)
            ls_losses[i] = l_star; ls_alphas[i] = a_star

        # 3. อัปเดตค่าน้ำหนักหลัก (w_current)
        idx_best = np.argmin(ls_losses)
        w_cand   = w_current + ls_alphas[idx_best] * directions[idx_best]

        if ls_losses[idx_best] < best_loss:
            best_loss = ls_losses[idx_best]
            w_current = w_cand.copy()
            best_w    = w_current.copy()

        hist_te.append(compute_acc(best_w, Xte, yte))
        hist_loss.append(best_loss)

        # 4. CMA-ES Update (สอนให้ระบบรู้ว่าทิศทางไหนรุ่ง)
        top_k = pop_size // 4
        idx_sorted = np.argsort(ls_losses)
        top_dirs = directions[idx_sorted[:top_k]]

        mean_d = np.mean(top_dirs, axis=0)
        centered = top_dirs - mean_d
        cov_d = (centered.T @ centered) / (top_k - 1) + np.eye(N_WEIGHTS) * 1e-4

    return best_w, hist_te, hist_loss, time.time()-t0

print("Cell 6: Proposed Novel Algorithms (including DELCMASE 🌟) loaded.")
```

```python
# ==========================================
# CELL 7: Benchmark Execution & Visualization
# ==========================================
import textwrap

# ── กำหนดรายชื่ออัลกอริทึมและสีประจำตัว ─────────────────────────
ALGO_DEFS = [
    # (ชื่อ, ฟังก์ชัน, สี, เส้นขอบ)
    ("Backprop",      run_backprop,         "#2196F3", ":"),   # Blue
    ("Standard GA",   standard_ga,          "#F44336", "--"),  # Red
    ("Standard CMA-ES", official_cma_es,      "#8D6E63", "--"),  # Brown
    ("GI-GA",         gradient_informed_ga, "#FF9800", "-."),  # Orange
    ("LAME v1",       lame_v1,              "#4CAF50", "-."),  # Green
    ("MEGA",          mega,                 "#00BCD4", "-"),   # Cyan
    ("DELS",          dels,                 "#9C27B0", "-"),   # Purple
    ("DELCMASE 🌟",   delcmase,             "#E91E63", "-"),   # Pink/Magenta (Hero)
]

def run_all_benchmarks(datasets):
    summary = {}

    for ds_name, (Xtr, Xte, ytr, yte) in datasets.items():
        print(f"\n{'='*70}")
        print(f" 🚀 Dataset: {ds_name} | Train {Xtr.shape} | Test {Xte.shape}")
        print(f"{'='*70}")
        summary[ds_name] = {}

        for algo_name, fn, _, _ in ALGO_DEFS:
            accs, times, hists = [], [], []
            print(f"  [{algo_name}] Running {N_RUNS} seeds...", end=" ")

            for seed in range(N_RUNS):
                if algo_name == "Backprop":
                    acc, t, lc = fn(Xtr, ytr, Xte, yte, seed=seed)
                    accs.append(acc); times.append(t)
                    padded = list(lc) + [lc[-1]]*(GENERATIONS-len(lc)) if lc else [np.nan]*GENERATIONS
                    hists.append(padded[:GENERATIONS])
                else:
                    _, hist_te, _, t = fn(Xtr, ytr, Xte, yte, seed=seed)
                    accs.append(hist_te[-1]); times.append(t)
                    hists.append(hist_te)

            mean_acc = np.mean(accs)
            std_acc  = np.std(accs)
            mean_time = np.mean(times)

            summary[ds_name][algo_name] = {
                "mean":   mean_acc,
                "std":    std_acc,
                "time":   mean_time,
                "hists":  np.array(hists),
            }
            print(f"✅ {mean_acc:.3f} ± {std_acc:.3f} | ⏱️ {mean_time:.1f}s")

    return summary

def plot_comprehensive_results(summary, datasets):
    ds_names = list(datasets.keys())
    n_ds     = len(ds_names)

    # สร้าง Canvas กราฟ 3 แถว
    fig, axes = plt.subplots(3, n_ds, figsize=(5*n_ds, 15))
    fig.suptitle("Evolutionary Algorithms vs Backpropagation\nHighlighting the novel DELCMASE 🌟",
                 fontsize=16, fontweight='bold', y=1.02)

    for col_i, ds_name in enumerate(ds_names):
        ds_res = summary[ds_name]

        # ── แถว 1: Convergence Curve (ความเร็วในการลู่เข้า) ──
        ax = axes[0, col_i]
        for algo_name, _, color, ls in ALGO_DEFS:
            r = ds_res[algo_name]
            mean_hist = r["hists"].mean(axis=0)
            std_hist  = r["hists"].std(axis=0)
            x = np.arange(len(mean_hist))

            # ทำให้เส้นของ DELCMASE หนาและชัดเจนกว่าเพื่อน
            lw = 2.5 if "DELCMASE" in algo_name else 1.2
            alpha_fill = 0.15 if "DELCMASE" in algo_name else 0.05

            ax.plot(x, mean_hist, color=color, linestyle=ls, linewidth=lw, label=f"{algo_name} ({r['mean']:.3f})")
            ax.fill_between(x, mean_hist-std_hist, mean_hist+std_hist, color=color, alpha=alpha_fill)

        ax.set_title(f"[{ds_name}] Convergence", fontsize=12)
        ax.set_xlabel("Generation / Epoch"); ax.set_ylabel("Test Accuracy")
        ax.legend(fontsize=7, loc="lower right")
        ax.grid(True, alpha=0.3)
        ax.set_ylim(0.4, 1.05)

        # ── แถว 2: Accuracy Bar Chart พร้อม Execution Time ──
        ax = axes[1, col_i]
        algo_names = [a[0] for a in ALGO_DEFS]
        means = [ds_res[a]["mean"] for a in algo_names]
        stds  = [ds_res[a]["std"]  for a in algo_names]
        clrs  = [a[2] for a in ALGO_DEFS]
        xs    = np.arange(len(algo_names))

        bars = ax.bar(xs, means, yerr=stds, capsize=4, color=clrs, alpha=0.85, edgecolor='black')
        ax.set_title(f"[{ds_name}] Accuracy (Mean ± SD)", fontsize=12)
        ax.set_xticks(xs); ax.set_xticklabels(algo_names, rotation=35, ha='right', fontsize=9)
        ax.set_ylim(max(0.0, min(means)-0.1), min(1.05, max(means)+0.1))

        # แปะค่าความแม่นยำ และ เวลาประมวลผลไว้บนแท่ง
        for bar, m, t in zip(bars, means, [ds_res[a]["time"] for a in algo_names]):
            ax.text(bar.get_x()+bar.get_width()/2, bar.get_height() + 0.01,
                    f"{m:.3f}\n({t:.1f}s)", ha='center', va='bottom', fontsize=8)
        ax.grid(axis='y', alpha=0.3)

        # ── แถว 3: Decision Boundary (DELCMASE vs Backprop) ──
        ax = axes[2, col_i]
        Xtr, Xte, ytr, yte = datasets[ds_name]

        # เทรน Backprop และ DELCMASE ด่วนๆ 1 รอบเพื่อวาดเส้น
        bp_mlp = MLPClassifier(hidden_layer_sizes=(16,16), activation='relu', solver='adam', max_iter=200, random_state=SEED)
        bp_mlp.fit(Xtr, ytr)

        # รัน DELCMASE 1 รอบแบบรวดเร็ว
        delc_w, _, _, _ = delcmase(Xtr, ytr, Xte, yte, seed=SEED)

        h = 0.05
        x0, x1 = Xte[:,0].min()-0.5, Xte[:,0].max()+0.5
        y0, y1 = Xte[:,1].min()-0.5, Xte[:,1].max()+0.5
        xx, yy = np.meshgrid(np.arange(x0,x1,h), np.arange(y0,y1,h))
        grid   = np.c_[xx.ravel(), yy.ravel()]

        Z_bp = (bp_mlp.predict_proba(grid)[:,1] >= 0.5).astype(int).reshape(xx.shape)
        Z_dc = (nn_forward(grid, delc_w) >= 0.5).astype(int).reshape(xx.shape)

        # วาดพื้นหลังด้วย DELCMASE (สีชมพู/ฟ้า)
        ax.contourf(xx, yy, Z_dc, alpha=0.2, cmap='RdBu')
        # วาดเส้นทับด้วย Backprop (เส้นประ)
        ax.contour(xx, yy, Z_bp, levels=[0.5], colors='navy', linewidths=1.5, linestyles='--')
        ax.scatter(Xte[:,0], Xte[:,1], c=yte, cmap='bwr', edgecolors='k', s=20, linewidths=0.5)

        ax.set_title(f"Boundary: DELCMASE (fill) vs Backprop (dashed)", fontsize=10)
        ax.set_xlabel("Feature 1"); ax.set_ylabel("Feature 2")

    plt.tight_layout()
    plt.savefig('delcmase_benchmark_results.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("\n🎉 พล็อตผลลัพธ์สำเร็จ! เซฟรูปภาพเป็น 'delcmase_benchmark_results.png' เรียบร้อยครับ")

# ==========================================
# 🌟 รันระบบทั้งหมด
# ==========================================
if __name__ == "__main__":
    summary_results = run_all_benchmarks(DATASETS)
    plot_comprehensive_results(summary_results, DATASETS)
```
